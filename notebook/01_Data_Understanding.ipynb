#### Placement Status Visualization

The distribution of placement outcomes is visualized to provide a clear understanding of the proportion of placed and non-placed students in the training dataset.

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=train_df, x="Placement_Status")

plt.title("Placement Status Distribution")
plt.xlabel("Placement Status")
plt.ylabel("Number of Students")
plt.show()


plt.figure(figsize=(7, 5))
ax = sns.countplot(data=train_df, x="Placement_Status")
total = len(train_df)

for bar in ax.patches:
    count = bar.get_height()
    percentage = (count / total) * 100
    
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        count + 500,
        f"{int(count):,}\n({percentage:.2f}%)",
        ha="center",
        va="bottom",
        fontsize=10
    )

plt.title("Placement Status Distribution")
plt.xlabel("Placement Status")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()
```
